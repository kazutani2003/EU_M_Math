8-7-1サポートベクターマシンのモデル構築

```python
from sklearn.datasets import load_breast_cancer
from sklearn.svm import LinearSVC
from sklearn.model_selection import train_test_split
cancer =load_breast_cancer()
x_train,x_test,y_train,y_test = train_test_split(
    cancer.data,cancer.target,stratify=cancer.target,random_state=0)
model = LinearSVC()
model.fit(x_train,y_train)

print('正解率(train):{:.3f}'.format(model.score(x_train,y_train)))
print('正解率(test):{:.3f}'.format(model.score(x_test,y_test)))
```

総合問題８－１

・回帰
意味： 「売上」や「気温」のように、連続した数値を予想すること。

・分類
意味： 「合格・不合格」「犬・猫」のように、グループ（カテゴリ）を分けること。


・重回帰分析
意味： 複数の原因（データ）から、1つの結果（数値）を予測する王道のやり方。

・ロジスティック回帰
意味： 名前に「回帰」とついていますが、実際は「分類」に使う手法。「あるイベントが起きる確率（0〜100%）」を計算して、最終的にAかBかを分類する。

・正規化
意味：過学習を防ぐための機能。

・リッジ回帰
意味： 正規化の種類のひとつ。すべての原因（データ）をバランスよく行う。

・ラッソ回帰
意味： 正規化の種類のひとつ。関係なさそうな原因の力を完全にゼロにする。結果として、本当に大事な原因だけが残る。

・決定木
意味： 「Yes / No」の質問を繰り返して、データをイエス・ノーの木（ツリー）のようなルートで分類していく手法。

・エントロピー
意味： データの中身の「ごちゃ混ぜ度（不純度）」を表す数値です。

・情報利得
意味： 決定木で質問をした結果、どれだけデータがスッキリ片付いたかの度合い。AIはこの情報利得が一番大きくなるような良い質問を探す。

・k-NN法
意味： 迷ったら「近くにある多数派に合わせる」という、一番直感的な分類方法です。

・SVM
意味： 2つのグループの境界線を引くとき、両方のグループから一番離れた、最もフェアで安全な境界線を引く手法。

・ノーフリーランチ
意味： 「どんな問題に対しても、常に最強で完璧なAIの手法なんて存在しない」という格言・定理。
