```python
import numpy as np
import random

sample_names = np.array(['a','b','c','d','a'])
np.random.seed(0)
data = np.random.randn(5,5)

print(sample_names)
print(data)
```

```python
new_data = np.delete(sample_names, 2)
new_data = np.delete(data ,2 ,axis=1)
print(new_data)
```
